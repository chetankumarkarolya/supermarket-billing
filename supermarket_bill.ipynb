```python
# write  a programe to create a biling system at supermarket
while True:
    name = input("enter name ")
    total = 0
    
    while True:
        print("enter the amount and quantity")
        quantity = float(input("enter quantity"))
        amount = float(input("enter amount"))
        total = total+amount*quantity
        
        repeat = input("can you repeat the order y/n")
        if repeat == "n" or repeat == "N":
            break
    
    print("-"*40)
    print("Name:",name)
    print("Amount to be paid :",total)
    print("-"*40)
    print("*"*8,"Happy Shopping","*"*8)

    repeat1 = input("do you want to go to next customer? (yes/no) :")
    if repeat1 == "no" or repeat1 == "No":
        break
```
